# **instrukcje warunkowe:**

```python
jestem_glodny = True
jestem_w_dobrym_nastroju = True

if jestem_glodny and jestem_w_dobrym_nastroju:
  print("będe jadł i oglądał film")
elif jestem_glodny and not jestem_w_dobrym_nastroju:
  print("będe jadł")
elif not jestem_glodny and jestem_w_dobrym_nastroju:
  print("będe oglądał film bez jedzenia")
else:
    print("nic nie zrobie")
```
