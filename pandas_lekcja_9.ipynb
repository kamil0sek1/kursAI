```python
import pandas as pd

df = pd.read_excel("/content/Gigathon_Wyniki.xlsx")

df
```

```python
print(df.head(5))
print(df.tail(5))
```

```python
df["Wynik procentowy"] = df['Punktacja']/120*100
df
```

```python
df.dropna(inplace=True)
df
```

```python
df.sort_values(by="Punktacja", ascending = False, inplace=True)
df
```

```python
df["Zakwalifikowany"] = df["Wynik procentowy"] > 70

df
```

```python
df.groupby([ "Zakwalifikowany", "Wiek"])[["Punktacja","Wynik procentowy"]].agg(["count","min","max","mean","median"])
#
```

```python
df_zak  = df[df["Zakwalifikowany"]]
df_zak.drop("Zakwalifikowany", axis=1, inplace=True)
df_zak.to_excel("zakwalifikowani.xlsx")

df



# nie_df_zak  = df[-df["Zakwalifikowany"]]
# nie_df_zak.drop("Zakwalifikowany", axis=1, inplace=True)
# nie_df_zak.to_excel("nie_zakwalifikowani.xlsx")
# df
```
